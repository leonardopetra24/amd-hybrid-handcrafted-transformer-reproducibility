# Portable preprocessing and assisted ROI
This notebook is a clean wrapper around the reconstructed modules. The automatic preprocessing is deterministic. ROI placement is operator-assisted; historical crop coordinates were not retained, so exact pixel-level replay of the original manual crops is not claimed.

```
from pathlib import Path
from src.preprocessing import preprocess_fundus
from src.assisted_roi_crop import assisted_crop

RAW_IMAGE = Path('path/to/raw_image.jpg')
PREPROCESSED_OUT = Path('work/preprocessed/example.png')
ROI_OUT = Path('work/roi/example.png')
```

```
import cv2
img = cv2.imread(str(RAW_IMAGE))
if img is None:
    raise FileNotFoundError(RAW_IMAGE)
pre = preprocess_fundus(img)
PREPROCESSED_OUT.parent.mkdir(parents=True, exist_ok=True)
cv2.imwrite(str(PREPROCESSED_OUT), pre)
print(PREPROCESSED_OUT)
```

```
# Interactive step; requires a desktop OpenCV GUI.
ROI_OUT.parent.mkdir(parents=True, exist_ok=True)
assisted_crop(PREPROCESSED_OUT, ROI_OUT, roi_size=224)
print(ROI_OUT)
```
